## Generate Figure 01 from the final statistical workbook

````
import subprocess, sys
from pathlib import Path

# Edit path if needed
final18 = Path("outputs/workbooks/LUNAR_Final18_Statistical_Comparison_Workbook.xlsx")

subprocess.run([
    sys.executable,
    "scripts/cross_dataset/generate_zscore_heatmaps.py",
    "--input", str(final18),
    "--output", "outputs/figures/Figure_01_Zscore_Heatmaps.png",
], check=True)
````
